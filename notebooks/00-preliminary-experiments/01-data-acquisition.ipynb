# 01 - data acquisition (sample)

load sample data from `data/samples/`, parse filenames, verify structure.

```python
from pathlib import Path
import pandas as pd
from src.utils.helpers import parse_filename, assign_split

SAMPLES = Path("../../data/samples")
df = pd.read_csv(SAMPLES / "sample_labels.csv")
print(f"loaded {len(df)} sample files")
```

```python
print(df.head())
print(df.dtypes)
```

```python
print(df.groupby(['split','channel']).size().unstack(fill_value=0))
```

```python
print("actors per split:")
for s in ['train','val','test']:
    a = sorted(df[df['split']==s]['actor'].unique())
    print(f"  {s}: {a} ({len(a)} actors)")
```

```python
print("file exists check:")
all_ok = all(Path(p).exists() for p in df['filepath'])
print(f"  all {len(df)} files exist: {all_ok}")
```
